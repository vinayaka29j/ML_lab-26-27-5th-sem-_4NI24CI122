```python
import numpy as np
import matplotlib.pyplot as plt

# Given data
x = np.array([-2, -1, 0, 1, 2], dtype=float)
y = np.array([9, 2, 1, 6, 17], dtype=float)

print("X =", x)
print("Y =", y)
```

```python
# Design matrix for degree 2 polynomial
X = np.column_stack((np.ones(len(x)), x, x**2))

print("Design Matrix X:")
print(X)
```

```python
# Calculate X transpose
XT = X.T

# Calculate X^T X
XTX = XT @ X

# Calculate X^T y
XTy = XT @ y

print("X^T X:")
print(XTX)

print("\nX^T y:")
print(XTy)
```

```python
# Calculate beta = (X^T X)^(-1) X^T y
beta = np.linalg.inv(XTX) @ XTy

print("Beta coefficients:")
print(beta)
```

```python
# Predicted values
y_pred = X @ beta

print("Actual Y:    ", y)
print("Predicted Y: ", y_pred)
```

```python
# Calculate errors
errors = y - y_pred

print("Errors:")
print(errors)
```

```python
# Calculate MSE
mse = np.mean((y - y_pred)**2)

print("MSE =", mse)
```

```python
# Matrix method for SSE
residuals = y - X @ beta

SSE = residuals.T @ residuals
MSE = SSE / len(y)

print("SSE =", SSE)
print("MSE =", MSE)
```

```python
# Generate smooth x values
x_line = np.linspace(-2, 2, 200)

# Create design matrix for smooth x values
X_line = np.column_stack((
    np.ones(len(x_line)),
    x_line,
    x_line**2
))

# Predicted values for smooth curve
y_line = X_line @ beta

# Plot original data
plt.scatter(x, y, color="red", s=70, label="Actual Data")

# Plot polynomial
plt.plot(
    x_line,
    y_line,
    color="blue",
    linewidth=2,
    label="y = 1 + 2x + 3x²"
)

plt.xlabel("X")
plt.ylabel("Y")
plt.title("Degree-2 Polynomial Using Matrix Method")

plt.legend()
plt.grid(True)

plt.show()
```
